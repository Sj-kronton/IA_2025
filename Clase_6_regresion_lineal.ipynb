```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
data=pd.read_excel("/content/Salary.xlsx")
data.head()
```

```python
df=pd.DataFrame(data) #crear datafrme y buscar las variables dpeendientes e independientes
df.head()
```

```python
x=df["YearsExperience"] #variables dependientes e independientes
y=df["Salary"]#esta es la dependiente

plt.figure(figsize=(10,6))
plt.scatter(x,y)
plt.title("Salary vs YearsExperience")
plt.legend()
plt.grid(True)
plt.show()
```

```python
model=LinearRegression() #ajustar datos al modelo de regresion lineal
model.fit(x.values.reshape(-1, 1),y)

#generar las predicciones
y_pred=model.predict(x.values.reshape(-1, 1))

#hallar el R2
r2=r2_score(y,y_pred)
print("R2:",r2)
```

```python


plt.figure(figsize=(10,6))
plt.scatter(x,y, color="pink", label="Datos reales")
plt.plot(x,y_pred,color="red", linewidth=2, label="Regresionlineal")
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.title("Salary vs YearsExperience")
plt.legend()
plt.grid(True)
plt.show()

print(f"Ecuacion de la recta y= {model.coef_[0]:.2f}x+ {model.intercept_:2f}")
print(f"Coeficiente de Determinacion (R^2): {r2:.2f}")
```
